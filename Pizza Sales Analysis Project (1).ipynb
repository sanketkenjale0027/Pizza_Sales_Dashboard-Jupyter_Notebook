#### Daily Trend - Total Orders

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['day_name'] = df['order_date'].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

df['day_name'] = pd.Categorical(
    df['day_name'],
    categories=weekday_order,
    ordered=True
)

orders_by_day = df.groupby('day_name', observed=False)['order_id'].nunique()

ax = orders_by_day.plot(kind='bar', figsize=(8,5),
                        color='green', edgecolor='black')

plt.title("Total Orders by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

for i, val in enumerate(orders_by_day):
    plt.text(i, val + 20, str(val), ha='center',
             va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

#### Daily Trend - Total Revenue

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['day_name'] = df['order_date'].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

df['day_name'] = pd.Categorical(
    df['day_name'],
    categories=weekday_order,
    ordered=True
)

orders_by_day = df.groupby('day_name', observed=False)['total_price'].sum()

ax = orders_by_day.plot(kind='bar', figsize=(8,5),
                        color='purple', edgecolor='black')

plt.title("Total Revenue by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)

for i, val in enumerate(orders_by_day):
    plt.text(i, val + 20, str(val), ha='center',
             va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

#### Hourly Trend - Total Orders

```python
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S')

df['order_hour'] = df['order_time'].dt.hour

orders_by_hour = df.groupby('order_hour', observed=False)['order_id'].nunique()

ax = orders_by_hour.plot(
    kind='bar',
    figsize=(10, 5),
    color='maroon',
    edgecolor='black'
)

plt.title("Total Orders by Hour of Day")
plt.xlabel("Hour of Day (24-Hour Format)")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)

for i, val in enumerate(orders_by_hour):
    plt.text(
        i, val + 5, str(val),
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

####  Monthly Trend - Total Orders

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['month_name'] = df['order_date'].dt.month_name()

month_order = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

df['month_name'] = pd.Categorical(df['month_name'], categories=month_order, ordered=True)

orders_by_month = df.groupby('month_name', observed=False)['order_id'].nunique()

plt.figure(figsize=(10,5))

plt.fill_between(orders_by_month.index, orders_by_month.values, color="orange", alpha=0.6)
plt.plot(orders_by_month.index, orders_by_month.values, color="black", linewidth=2, marker='o')

plt.title("Total Orders by Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

for i, val in enumerate(orders_by_month):
    plt.text(i, val + 20, str(val), ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

####  % of Sales by Category

```python
category_sales = df.groupby('pizza_category')['total_price'].sum()

category_pct = category_sales / category_sales.sum() * 100

plt.figure(figsize=(7,7))
colors = plt.get_cmap('tab20').colors  

plt.pie(category_pct, labels=category_pct.index, autopct='%1.1f%%', startangle=90, colors=colors, wedgeprops={'edgecolor':'black', 'width':0.4})

plt.title("Percentage of Sales by Pizza Category")
plt.show()
```

####   % Sales by Pizza Size & Category

```python
sales_pivot = df.pivot_table(
    index='pizza_category',
    columns='pizza_size',
    values='total_price',
    aggfunc='sum',
    fill_value=0
)

sales_pct = sales_pivot / sales_pivot.sum().sum() * 100

plt.figure(figsize=(10,6))
sns.heatmap(sales_pct, annot=True, fmt=".1f", cmap="YlOrRd", linewidths=0.5)

plt.title("% of Sales by Pizza Category and Size")
plt.ylabel("Pizza Category")
plt.xlabel("Pizza Size")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------------------
# DATA PREPARATION
# ---------------------------------------

df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)
df['order_time'] = pd.to_datetime(df['order_time'], format='%H:%M:%S')

df['day_name'] = df['order_date'].dt.day_name()
df['order_hour'] = df['order_time'].dt.hour
df['month_name'] = df['order_date'].dt.month_name()

weekday_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df['day_name'] = pd.Categorical(
    df['day_name'], categories=weekday_order, ordered=True
)

df['month_name'] = pd.Categorical(
    df['month_name'], categories=month_order, ordered=True
)

# ---------------------------------------
# KPI CALCULATIONS
# ---------------------------------------

total_revenue = df['total_price'].sum()
total_pizzas = df['quantity'].sum()
total_orders = df['order_id'].nunique()

avg_order_value = total_revenue / total_orders
avg_pizzas_per_order = total_pizzas / total_orders

# ---------------------------------------
# CREATE DASHBOARD
# ---------------------------------------

fig, axes = plt.subplots(4, 3, figsize=(22, 25))

fig.suptitle(
    "PIZZA SALES ANALYSIS DASHBOARD",
    fontsize=24,
    fontweight='bold',
    y=0.99
)

# KPI SUMMARY
fig.text(
    0.5, 0.965,
    f"Total Revenue: ${total_revenue:,.2f}     |     "
    f"Total Pizzas Sold: {total_pizzas:,}     |     "
    f"Total Orders: {total_orders:,}     |     "
    f"Average Order Value: ${avg_order_value:.2f}     |     "
    f"Average Pizzas Per Order: {avg_pizzas_per_order:.2f}",
    ha='center',
    fontsize=12,
    fontweight='bold'
)

# ---------------------------------------
# 1. TOP 5 PIZZAS SOLD BY QUANTITY
# ---------------------------------------

pizzas_by_name = df.groupby('pizza_name')['quantity'].sum()
top5_quantity = pizzas_by_name.sort_values(ascending=False).head(5)

top5_quantity.plot(
    kind='bar',
    ax=axes[0, 0],
    color='grey',
    edgecolor='black'
)

axes[0, 0].set_title("Top 5 Pizzas Sold")
axes[0, 0].set_xlabel("Pizza Name")
axes[0, 0].set_ylabel("Total Pizzas Sold")
axes[0, 0].tick_params(axis='x', rotation=45)

for i, val in enumerate(top5_quantity):
    axes[0, 0].text(i, val + 2, str(val),
                    ha='center', fontweight='bold', fontsize=8)

# ---------------------------------------
# 2. TOP 5 PIZZAS BY NUMBER OF ORDERS
# ---------------------------------------

pizzas_by_orders = df.groupby('pizza_name')['order_id'].nunique()
top5_orders = pizzas_by_orders.sort_values(ascending=False).head(5)

top5_orders.plot(
    kind='bar',
    ax=axes[0, 1],
    color='grey',
    edgecolor='black'
)

axes[0, 1].set_title("Top 5 Pizzas Ordered")
axes[0, 1].set_xlabel("Pizza Name")
axes[0, 1].set_ylabel("Total Orders")
axes[0, 1].tick_params(axis='x', rotation=45)

for i, val in enumerate(top5_orders):
    axes[0, 1].text(i, val + 2, str(val),
                    ha='center', fontweight='bold', fontsize=8)

# ---------------------------------------
# 3. BOTTOM 5 PIZZAS BY REVENUE
# ---------------------------------------

pizzas_revenue = df.groupby('pizza_name')['total_price'].sum()
bottom5 = pizzas_revenue.sort_values(ascending=True).head(5)

bottom5.plot(
    kind='bar',
    ax=axes[0, 2],
    color='brown',
    edgecolor='black'
)

axes[0, 2].set_title("Bottom 5 Pizzas by Revenue")
axes[0, 2].set_xlabel("Pizza Name")
axes[0, 2].set_ylabel("Total Revenue")
axes[0, 2].tick_params(axis='x', rotation=45)

for i, val in enumerate(bottom5):
    axes[0, 2].text(i, val + 2, f"${val:,.0f}",
                    ha='center', fontsize=8)

# ---------------------------------------
# 4. TOP 5 PIZZAS BY REVENUE
# ---------------------------------------

top5_revenue = pizzas_revenue.sort_values(ascending=False).head(5)

top5_revenue.plot(
    kind='bar',
    ax=axes[1, 0],
    color='grey',
    edgecolor='black'
)

axes[1, 0].set_title("Top 5 Pizzas by Revenue")
axes[1, 0].set_xlabel("Pizza Name")
axes[1, 0].set_ylabel("Total Revenue")
axes[1, 0].tick_params(axis='x', rotation=45)

for i, val in enumerate(top5_revenue):
    axes[1, 0].text(i, val + 2, f"${val:,.0f}",
                    ha='center', fontsize=8)

# ---------------------------------------
# 5. TOP 5 INGREDIENTS
# ---------------------------------------

ingredient = (
    df['pizza_ingredients']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(5)
)

ingredient.sort_values().plot(
    kind='barh',
    ax=axes[1, 1],
    color='teal',
    edgecolor='black'
)

axes[1, 1].set_title("Top 5 Ingredients")
axes[1, 1].set_xlabel("Ingredient Count")
axes[1, 1].set_ylabel("Ingredients")

# ---------------------------------------
# 6. DAILY TREND - TOTAL ORDERS
# ---------------------------------------

orders_by_day = df.groupby(
    'day_name', observed=False
)['order_id'].nunique()

orders_by_day.plot(
    kind='bar',
    ax=axes[1, 2],
    color='green',
    edgecolor='black'
)

axes[1, 2].set_title("Total Orders by Day of Week")
axes[1, 2].set_xlabel("Day of Week")
axes[1, 2].set_ylabel("Number of Orders")
axes[1, 2].tick_params(axis='x', rotation=45)

for i, val in enumerate(orders_by_day):
    axes[1, 2].text(i, val + 20, str(val),
                    ha='center', fontsize=8, fontweight='bold')

# ---------------------------------------
# 7. DAILY TREND - TOTAL REVENUE
# ---------------------------------------

revenue_by_day = df.groupby(
    'day_name', observed=False
)['total_price'].sum()

revenue_by_day.plot(
    kind='bar',
    ax=axes[2, 0],
    color='purple',
    edgecolor='black'
)

axes[2, 0].set_title("Total Revenue by Day of Week")
axes[2, 0].set_xlabel("Day of Week")
axes[2, 0].set_ylabel("Total Revenue")
axes[2, 0].tick_params(axis='x', rotation=45)

for i, val in enumerate(revenue_by_day):
    axes[2, 0].text(i, val + 20, f"${val:,.0f}",
                    ha='center', fontsize=8)

# ---------------------------------------
# 8. HOURLY TREND - TOTAL ORDERS
# ---------------------------------------

orders_by_hour = df.groupby(
    'order_hour', observed=False
)['order_id'].nunique()

orders_by_hour.plot(
    kind='bar',
    ax=axes[2, 1],
    color='maroon',
    edgecolor='black'
)

axes[2, 1].set_title("Total Orders by Hour of Day")
axes[2, 1].set_xlabel("Hour of Day (24-Hour Format)")
axes[2, 1].set_ylabel("Number of Orders")
axes[2, 1].tick_params(axis='x', rotation=0)

for i, val in enumerate(orders_by_hour):
    axes[2, 1].text(i, val + 5, str(val),
                    ha='center', fontsize=7, fontweight='bold')

# ---------------------------------------
# 9. MONTHLY TREND - TOTAL ORDERS
# ---------------------------------------

orders_by_month = df.groupby(
    'month_name', observed=False
)['order_id'].nunique()

axes[2, 2].fill_between(
    range(len(orders_by_month)),
    orders_by_month.values,
    color='orange',
    alpha=0.6
)

axes[2, 2].plot(
    range(len(orders_by_month)),
    orders_by_month.values,
    color='black',
    linewidth=2,
    marker='o'
)

axes[2, 2].set_title("Total Orders by Month")
axes[2, 2].set_xlabel("Month")
axes[2, 2].set_ylabel("Number of Orders")
axes[2, 2].set_xticks(range(len(orders_by_month)))
axes[2, 2].set_xticklabels(
    orders_by_month.index, rotation=45
)

for i, val in enumerate(orders_by_month):
    axes[2, 2].text(i, val + 20, str(val),
                    ha='center', fontsize=8, fontweight='bold')

# ---------------------------------------
# 10. PERCENTAGE OF SALES BY CATEGORY
# ---------------------------------------

category_sales = df.groupby('pizza_category')['total_price'].sum()

category_pct = category_sales / category_sales.sum() * 100

axes[3, 0].pie(
    category_pct,
    labels=category_pct.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=plt.get_cmap('tab20').colors,
    wedgeprops={'edgecolor': 'black', 'width': 0.4}
)

axes[3, 0].set_title("Percentage of Sales by Pizza Category")

# ---------------------------------------
# 11. SALES BY PIZZA SIZE AND CATEGORY
# ---------------------------------------

sales_pivot = df.pivot_table(
    index='pizza_category',
    columns='pizza_size',
    values='total_price',
    aggfunc='sum',
    fill_value=0
)

sales_pct = sales_pivot / sales_pivot.sum().sum() * 100

sns.heatmap(
    sales_pct,
    annot=True,
    fmt='.1f',
    cmap='YlOrRd',
    linewidths=0.5,
    ax=axes[3, 1]
)

axes[3, 1].set_title("% of Sales by Pizza Category and Size")
axes[3, 1].set_ylabel("Pizza Category")
axes[3, 1].set_xlabel("Pizza Size")

# ---------------------------------------
# 12. TOTAL PIZZAS SOLD BY CATEGORY
# ---------------------------------------

pizzas_by_category = df.groupby(
    'pizza_category'
)['quantity'].sum()

pizzas_by_category.plot(
    kind='bar',
    ax=axes[3, 2],
    color='red',
    edgecolor='black'
)

axes[3, 2].set_title("Total Pizzas Sold by Pizza Category")
axes[3, 2].set_xlabel("Pizza Category")
axes[3, 2].set_ylabel("Total Pizzas Sold")
axes[3, 2].tick_params(axis='x', rotation=45)

for i, val in enumerate(pizzas_by_category):
    axes[3, 2].text(i, val + 5, str(val),
                    ha='center', fontsize=8, fontweight='bold')

# ---------------------------------------
# FINAL DASHBOARD LAYOUT
# ---------------------------------------

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
